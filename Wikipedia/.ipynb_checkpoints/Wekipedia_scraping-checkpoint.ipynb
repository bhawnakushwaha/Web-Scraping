## Article scrapping from wikipedia

```python
import requests
from bs4 import BeautifulSoup
```

```python
link = 'https://en.wikipedia.org/wiki/Dehradun'

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}  # in http requests, headers  are extra information send along with request, 
# which indicate that its a normal web browser which prevent blocking specially sites like wikipedia, filpcart etc for web scrapping.

res = requests.get(link, headers=headers) 
```

```python
res   # 403, because for wikipedia we have no permission to access the requests. 
```

```python
soup = BeautifulSoup(res.text, "html.parser")
```

```python
# soup.text
```

```python
heading = soup.find('h1').text
```

```python
heading
```

```python
corpus = ''

for p in soup.find_all('p'):
    corpus += p.text
    corpus += '\n'

corpus = corpus.strip()
```

```python
# print(corpus)
```

```python
# for removing the citation.
for i in range(1,155):
    corpus = corpus.replace('['+str(i)+']', '')
```

```python
# print(corpus)
```

```python
# saving into file
fd = open(heading + '.txt', 'w',encoding='utf-8')
fd.write(corpus)
fd.close() 
```
